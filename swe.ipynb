```python
import os
import pandas as pd

# 1. Locate and load the CSV from Downloads
downloads_path = os.path.expanduser("~/Downloads")
file_name = "report.csv"  # <-- Change this to your file name
file_path = os.path.join(downloads_path, file_name)

df = pd.read_csv(file_path)

# 2. Generate and print a text summary report
print("=" * 40)
print(f"       DATA REPORT: {file_name}")
print("=" * 40)
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
print("\n--- Column Names & Data Types ---")
print(df.dtypes)
print("\n--- Missing Values per Column ---")
print(df.isnull().sum())
print("=" * 40)

# 3. Display the interactive table preview
print("\nPreview of the Data (First 5 rows):")
display(df.head())
```
